### Pandas-DataFrame And Series 

```python
## series 

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series\n",series);
print(type(series))
```

```python
## create a series from dictionary
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
## Dataframe
## create a DataFrame from a dicionary of list

data={
    'Name':['Abdul','Rahman','Ayyan'],
    'Age':[21,23,20],
    'City':['Bangalore','New York','florida']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
print(df)
```

```python
## create a data frame from a list of dictionaries
data=[
    {'Name':'Abdul','Age':32,'City':'Bangalore'},
    {'Name':'Rahman','Age':22,'City':'Bangalore'},
    {'Name':'Ayyan','Age':42,'City':'Bangalore'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
